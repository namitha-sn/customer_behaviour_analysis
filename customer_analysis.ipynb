```python
df['Review Rating']=df.groupby('Category')['Review Rating'].transform(lambda x:x.fillna(x.median()))
```

```python
df.isna().sum()
```

```python
df.columns = (
    df.columns
      .str.lower()
      .str.replace(' ', '_', regex=False)   # remove existing underscores
      .str.strip()
)
```

```python
df.columns
```

```python
#create a column age_group
labels=['Young adult','Adult','Middle-aged','Senior']
df['agegroup']=pd.qcut(df['age'],q=4,labels=labels)
```

```python
df[['age','agegroup']].head(10)
```

```python
#create column purchase_frequency_days it is in category
frequency_map={
    'Fortnightly':14,
    'Weekly':7,
    'monthly':30,
    'Quarterly':90,
    'Bi-weekly':14,
    'Annually':365,
    'Every 3 Months':90
}
df['purchase_freq']=df['frequency_of_purchases'].map(frequency_map)
```

```python
from sqlalchemy import create_engine
username="root"
password="root"
host="localhost"
port="3306"
database="customer_shopping"

engine=create_engine(f"mysql+mysqlconnector://{username}:{password}@{host}:{port}/{database}")
```

```python
table_name="customer_data"
df.to_sql(table_name,engine,if_exists="replace",index=False)
print("data loaded")
```

```python
import pandas as pd
query="select * from customer_data"
df_sql=pd.read_sql(query,engine)
print(df_sql)
```
